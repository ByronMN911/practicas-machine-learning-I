````python
# ============================
# Stack Plot (Gráfico de áreas apiladas)
# ============================

# Valores del eje X (años)
x = [2008, 2009, 2010, 2011, 2012]

# Porcentaje de participación de cada sistema operativo móvil por año
iOS = [35, 25, 20, 17, 18]
Android = [0, 15, 35, 45, 60]
WindowsPhone = [1, 3, 5, 5, 2]
Others = [64, 57, 40, 33, 20]

# Apila todas las listas verticalmente para formar una matriz de NumPy.
# Cada fila representa una categoría (sistema operativo).
# Esta matriz puede utilizarse como entrada para stackplot.
y = np.vstack([iOS, Android, WindowsPhone, Others])

# Muestra la matriz resultante en la consola.
print(y)

# Etiquetas que aparecerán en la leyenda del gráfico.
labels = ["iOS", "Android", "WindowsPhone", "Others"]

# Crea una figura y un eje donde se dibujará el gráfico.
fig, ax = plt.subplots()

# Dibuja el gráfico de áreas apiladas(stackplot).
# Cada lista representa una categoría cuyos valores se acumulan
# sobre la categoría anterior para mostrar la composición del total.
ax.stackplot(x, iOS, Android, WindowsPhone, Others, labels=labels)

# También es posible generar el mismo gráfico utilizando
# la matriz 'y' creada con np.vstack().
# ax.stackplot(x, y, labels=labels)

# Define las posiciones de las marcas (ticks) del eje X.
ax.set_xticks(x)

# Muestra la leyenda en la esquina superior izquierda.
ax.legend(loc='upper left')

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

````python
# ============================
# Scatter Plot (Diagrama de dispersión)
# ============================

# Importa la función rand para generar números aleatorios entre 0 y 1.
from numpy.random import rand

# Crea una figura y un eje donde se dibujará el gráfico.
fig, ax = plt.subplots()

# Número de puntos que tendrá cada grupo de datos.
n = 700

# Recorre una lista de colores para crear tres conjuntos de puntos:
# uno rojo, uno verde y uno azul.
for color in ['red', 'green', 'blue']:

    # Genera dos arreglos de tamaño n con valores aleatorios:
    # x contiene las coordenadas del eje X.
    # y contiene las coordenadas del eje Y.
    x, y = rand(2, n)

    # Genera un tamaño aleatorio para cada punto.
    # Esto hace que algunos marcadores sean más grandes que otros.
    scale = 25.0 * rand(n)

    # Dibuja el conjunto de puntos.
    ax.scatter(
        x, y,                 # Coordenadas de los puntos
        c=color,              # Color del grupo
        s=scale,              # Tamaño de cada punto
        label=color,          # Etiqueta para la leyenda
        alpha=0.3,            # Transparencia (30%)
        edgecolors='none'     # Sin borde alrededor de los puntos
    )

# Muestra la leyenda con los nombres de los grupos.
ax.legend()

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

# frame.plot()

```
frame.plot(x,y,label=<Series>,ax,style,kind,xticks,yticks,title,subplots)
```

* (solo para Series) label: referencia para la leyenda
* ax: subfigura en la que dibujar los datos
* style: estilo de la línea
* kind: tipo de gráfico ('bar', 'pie', 'hist', 'area' 'line', 'barh', 'density', 'kde').
* xticks y yticks: valores en los ejes X e Y
* title: string como título
* (solo para DataFrame) subplots: bool si se desean subfiguras separadas para cada columna.
* (solo para DataFrame) 'x' e 'y' se pueden utilizar para seleccionar columnas para cada eje

````python
import seaborn as sns
````

````python
# ============================
# Swarm Plot (Gráfico de enjambre)
# ============================

# Carga el conjunto de datos de ejemplo "flights" incluido en Seaborn.
# Este dataset contiene el número de pasajeros de vuelos por mes y año.
flights_data = sns.load_dataset("flights")

# Muestra el DataFrame completo (opcional).
# display(flights_data)

# Construye un gráfico de enjambre (Swarm Plot).
# - x: representa la variable categórica (mes).
# - y: representa la variable numérica (cantidad de pasajeros).
# - data: conjunto de datos que contiene ambas variables.
#
# El Swarm Plot distribuye automáticamente los puntos para evitar que
# se superpongan, permitiendo visualizar la distribución de los datos
# dentro de cada categoría.
sns.swarmplot(x="month", y="passengers", data=flights_data)

# Rota las etiquetas del eje X 90 grados para mejorar su legibilidad.
plt.xticks(rotation=90)

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

**Compatible con pandas**

````python
# Load in data
tips = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")
tips.sample(5)
````

````python
# ============================
# Bar Plot (Gráfico de barras)
# ============================

# Crea un gráfico de barras utilizando el conjunto de datos "tips".
#
# - x='sex': agrupa las barras por sexo (Male y Female).
# - y='tip': calcula y muestra el promedio de la propina para cada grupo.
# - hue='time': divide cada barra según el momento del día (Lunch o Dinner),
#   permitiendo comparar ambas categorías dentro de cada sexo.
# - data=tips: conjunto de datos utilizado para construir el gráfico.
#
# De forma predeterminada, Seaborn representa la media de cada grupo
# e incluye barras de error que muestran un intervalo de confianza
# de la estimación de la media.
#
# El parámetro 'palette' (opcional) permite cambiar la paleta de colores
# utilizada para las barras.
sns.barplot(x='sex', y='tip', hue='time', data=tips)

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

````python
# ============================
# Box Plot (Diagrama de caja y bigotes)
# ============================

# Crea un diagrama de caja para analizar la distribución de las propinas
# según el día de la semana.
#
# - x='day': agrupa los datos por día (Thur, Fri, Sat y Sun).
# - y='tip': representa el valor de las propinas.
# - data=tips: conjunto de datos utilizado para construir el gráfico.
#
# El Box Plot permite visualizar:
# - La mediana (línea dentro de la caja).
# - El primer cuartil (Q1) y el tercer cuartil (Q3), que forman la caja.
# - Los bigotes, que representan la dispersión de los datos.
# - Los valores atípicos (outliers), que aparecen como puntos fuera de los bigotes.
sns.boxplot(x='day', y='tip', data=tips)

# Forma equivalente de crear el mismo gráfico especificando
# directamente las columnas del DataFrame.
# sns.boxplot(x=tips['day'], y=tips.tip)

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

## Distribución de un dataset

````python
# ============================
# Histograma con curva de densidad (Density Plot)
# ============================

# Crea un gráfico de distribución utilizando la variable 'tip'
# (propinas) del conjunto de datos 'tips'.
#
# - tips['tip']: datos numéricos que se representarán.
# - kde=True: agrega una curva de densidad (Kernel Density Estimate)
#   que muestra la distribución estimada de los datos de forma continua.
# - bins=10: divide los datos en 10 intervalos (contenedores o barras)
#   para construir el histograma.
# - rug=True: añade pequeñas marcas sobre el eje X que indican
#   la posición exacta de cada observación en el conjunto de datos.
sns.displot(tips['tip'], kde=True, bins=10, rug=True)

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````

````python
#scatter de clases, coloreando segun clase en matplotlib
iris = sns.load_dataset('iris')

# preparacion de los datos
data = []
colors = ["red", "green", "blue"]
groups = ["setosa", "versicolor", "virginica"]
columns = ["petal_length", "petal_width"]

for i in groups:
    # busqueda por clase definida
    data.append(iris[iris["species"] == i][columns].values)

#creacion de la grafica
fig = plt.figure()
ax = fig.add_subplot()

for count, d in enumerate(data):
    # separar las variables del grafico
    x, y = d[:,0], d[:,1]
    ax.scatter(x, y, alpha = 0.4, c = colors[count -1], edgecolors = 'none', s = 30, label = groups[count - 1])

#etiquetas
ax.set_xlabel(columns[0])
ax.set_ylabel(columns[1])

plt.legend(loc='best')
plt.title('Matplot scatter plot')
plt.show()
````

````python
# ============================
# Reg Plot (Gráfico de regresión)
# ============================

# Crea un gráfico de dispersión con una línea de regresión lineal.
#
# - x="total_bill": representa el valor total de la cuenta.
# - y="tip": representa el valor de la propina.
# - data=tips: conjunto de datos utilizado.
#
# El gráfico permite visualizar:
# - La relación (correlación) entre ambas variables.
# - La línea de regresión, que representa la tendencia o modelo
#   de predicción lineal de los datos.
# - La banda sombreada alrededor de la línea, que corresponde
#   al intervalo de confianza de la estimación (95% por defecto).
#
# Si los puntos están más cercanos a la línea de regresión,
# significa que el modelo lineal representa mejor los datos
# y, en general, sus predicciones serán más precisas.
sns.regplot(x="total_bill", y="tip", data=tips)

# Renderiza y muestra el gráfico en pantalla.
plt.show()
````
