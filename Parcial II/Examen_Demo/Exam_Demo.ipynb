```python
numbers = np.array([4, 4, 4, 3, 8, 1, 9, 6, 1, 5, 9, 4, 0, 2])
expected_output = np.array([False, True, True, False, False, False, False, False, True, False, True, True, False, False])
output = find_duplicates(numbers)   # This calls your function find_duplicates
print(output)
assert(np.array_equal(expected_output, output))  # This will fail if the result is not as expected
```

```python
# Your solution HERE !!!!
# Función que identifica los elementos duplicados de un arreglo
def find_duplicates(numbers):

    # Obtiene los índices de la primera aparición de cada valor único
    _, idx = np.unique(numbers, return_index=True)

    # Crea un arreglo de valores booleanos inicializado en False
    result = np.zeros_like(numbers, dtype=bool)

    # Marca como True las posiciones donde aparece cada valor por primera vez
    result[idx] = True

    # Invierte los valores booleanos:
    # - Las primeras apariciones pasan a False.
    # - Los elementos repetidos pasan a True.
    result = np.invert(result)

    # Devuelve el arreglo indicando qué elementos son duplicados
    return result
```

```python
# Función que identifica los elementos duplicados de un arreglo
def find_duplicates(numbers):

    # Conjunto para almacenar los valores que ya han aparecido
    visto = set()

    # Lista donde se guardará el resultado
    output = []

    # Recorre cada número del arreglo
    for value in numbers:

        # Si el número ya fue visto, es un duplicado
        if value in visto:
            output.append(True)

        # Si aparece por primera vez, se guarda en el conjunto
        else:
            visto.add(value)
            output.append(False)

    # Convierte la lista de resultados en un arreglo de NumPy
    return np.array(output)
```

```python
# Función que identifica los elementos duplicados de un arreglo
def find_duplicates(numeros):

    # Diccionario para registrar los valores que ya aparecieron
    dicci = {}

    # Crea un arreglo booleano inicializado en False
    resul = np.full(numeros.shape, False, dtype=bool)

    # Recorre el arreglo obteniendo el índice y el valor
    for idx, valor in enumerate(numeros):

        # Si el valor ya existe en el diccionario, es un duplicado
        if valor in dicci:
            resul[idx] = True

        # Si es la primera vez que aparece, se registra en el diccionario
        else:
            dicci[valor] = True

    # Devuelve un arreglo donde:
    # False = primera aparición
    # True = elemento duplicado
    return resul
```

## 03
- Resolución de dos sistemas de ecuaciones
    - $ A : \begin{cases} -9x +  4y = 20 \\ -7y + 16x = 80                         \end{cases}$
    
    - $ B : \begin{cases} x - 2y + 3z = 7 \\ 2x + y + z = 4 \\ -3x + 2y -2z = -10 \end{cases}$
- Verifica por código los resultados

```python
# Importa la librería para realizar comparaciones de números decimales
import math

# Importa el módulo de álgebra lineal de NumPy
import numpy.linalg as linalg

# ==========================
# Sistema de ecuaciones 2x2
# ==========================

# Matriz de coeficientes
a = np.array([
    [-9, 4],
    [16, -7]
])

# Vector de términos independientes
b = np.array([20, 80])

# Muestra la matriz y el vector
print(a)
print(b)

# Resuelve el sistema de ecuaciones Ax = b
# x contiene las soluciones de las incógnitas
x, y = linalg.solve(a, b)

# Imprime los valores obtenidos
print(f'x={x} y={y}')

# Verifica que las soluciones satisfacen la primera ecuación
print(math.isclose(-9 * x + 4 * y, 20))

# Verifica que las soluciones satisfacen la segunda ecuación
print(math.isclose(16 * x - 7 * y, 80))

# ==========================
# Sistema de ecuaciones 3x3
# ==========================

# Matriz de coeficientes
A2 = np.array([
    [1, -2, 3],
    [2, 1, 1],
    [-3, 2, -2]
])

# Vector de términos independientes
b2 = np.array([7, 4, -10])

# Resuelve el sistema de ecuaciones
x, y, z = np.linalg.solve(A2, b2)

# Imprime los valores obtenidos
print(f'x={x} y={y} z={z}')

# Verifica que la solución cumple la primera ecuación
print(math.isclose(1 * x - 2 * y + 3 * z, 7))

# Verifica que la solución cumple la segunda ecuación
print(math.isclose(2 * x + y + z, 4))

# Verifica que la solución cumple la tercera ecuación
print(math.isclose(-3 * x + 2 * y - 2 * z, -10))
```

## 07 - 1pt
- Escribir un generador `frange` que simule el comportamiento de la función `arange` de NumPy
- Esta función cada vez que se invoque debe devolver valores espaciados uniformemente a partir de un valor inicial. Se debe proveer a dicha función un valor inicial para la secuencia de números y un valor para el step
- Debe soportar valores númericos en coma flotante
- La función potencialmente se puede llamar un número infinito de veces
- Usando plain Python, sin ninguna librería

```python
l = np.arange(0,5)
print(l)
```

```python
# Función generadora que produce una secuencia infinita de números
def frange(init, step):

    # Se ejecuta indefinidamente hasta que el usuario detenga el generador
    while True:

        # yield convierte a la función en un generador.
        # Devuelve el valor actual sin finalizar la función,
        # por lo que en la siguiente llamada continuará desde este punto.
        yield init

        # Incrementa el valor según el paso indicado
        init += step
```

```python
# Importa la librería para comparar números de punto flotante
import math

# Valor inicial de la secuencia
init = 10.6

# Incremento entre cada número generado
step = 0.8

# Crea el generador llamando a la función frange
g = frange(init, step)

# Cantidad de valores que se desean obtener del generador
iters = 10

# Obtiene los primeros 10 valores del generador utilizando next()
numbers = [next(g) for n in range(iters)]

# Recorre los valores generados junto con su índice
for i, n in enumerate(numbers):

    # Comprueba que cada valor sea igual a:
    # valor_inicial + (paso × posición)
    # math.isclose() se utiliza para evitar errores de precisión
    # propios de los números decimales.
    print(f"{i}---{n}"+"\n")
    assert(math.isclose(n, init + (step * i)))

# Muestra la lista de valores generados
print(numbers)
```

- Crear una función que reciba una Serie de Pandas y devuelva esa misma serie, reemplazando los espacios en blanco por el carácter menos frecuente en dicha serie.

```python
serie = pd.Series(list('bba cabc faabba aacbbfe'))
# print(serie)
expected_output = pd.Series(list('bbaecabcefaabbaeaacbbfe'))
output = clean_whitespace_replace(serie)  # This calls your function clean_whitespace_replace
print(output)
assert(expected_output.equals(output))  # This will fail if result is not as expected
```

```python
from collections import Counter

# Función que reemplaza los espacios en blanco por el carácter menos frecuente
def clean_whitespace_replace(seri):

    # dropna():
    # Elimina los valores nulos (NaN) de la Serie para evitar que
    # interfieran en el conteo de caracteres.
    #
    # astype(str):
    # Convierte cada elemento de la Serie al tipo de dato string (cadena),
    # permitiendo aplicar operaciones de texto sobre todos los elementos.
    #
    # ''.join(...):
    # Une todos los elementos de la Serie en una única cadena de texto
    # sin utilizar ningún separador.
    texto = ''.join(seri.dropna().astype(str))

    # Counter():
    # Cuenta cuántas veces aparece cada carácter dentro de la cadena.
    #
    # La comprensión:
    # (i for i in texto if i != '')
    # recorre cada carácter de la cadena y lo envía al contador.
    frecu = Counter(i for i in texto if i != ' ')
    print(frecu)

    # min():
    # Obtiene el carácter cuya frecuencia es la menor.
    #
    # key=frecu.get indica que la comparación se realiza utilizando
    # la cantidad de apariciones de cada carácter almacenada en Counter.
    mefrecu = min(frecu, key=frecu.get)

    # astype(str):
    # Convierte nuevamente todos los elementos de la Serie a cadenas.
    #
    # apply():
    # Recorre cada elemento de la Serie y aplica la función indicada.
    #
    # lambda s:
    # Función anónima que recibe cada elemento de la Serie.
    #
    # replace(' ', mefrecu):
    # Sustituye cada espacio en blanco (' ') por el carácter
    # menos frecuente encontrado anteriormente.
    seridif = seri.astype(str).apply(
        lambda s: s.replace(' ', mefrecu)
    )

    # Devuelve la Serie con los espacios reemplazados
    return seridif
```
