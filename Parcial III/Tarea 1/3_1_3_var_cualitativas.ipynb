```python
# Creación aleatoria de datos
import numpy as np
import random

#Esta función nos ayuda a obtener los datos de forma aleatoria, esto
#permite que los datos no se obtenga de forma secuencial
#El número 42 son las veces que reordenamos el conjunto de datos
random.seed(42)
```

```python
# Creamos una lista vacía donde se almacenarán los niveles de estudios
nivel_estudios = []

# Cargamos 110 veces el texto "Secundaria" a la lista
nivel_estudios.extend(['Secundaria'] * 110)

# Cargamos 172 veces "Bachillerato"
nivel_estudios.extend(['Bachillerato'] * 172)

# Cargamos 145 veces "Grado"
nivel_estudios.extend(['Grado'] * 145)

# Cargamos 52 veces "Master"
nivel_estudios.extend(['Master'] * 52)

# Cargamos 21 veces "Doctorado"
nivel_estudios.extend(['Doctorado'] * 21)

# Mezclamos aleatoriamente el orden de todos los elementos de la lista.
# Esto se llama una "permutación": los elementos siguen siendo los mismos
# y aparecen el mismo número de veces, pero en un orden completamente aleatorio.
nivel_estudios = np.random.permutation(nivel_estudios)
```

```python
print(nivel_estudios)
```

```python
# TABLA DE FRECUENCIA SIMPLE

# Importa la biblioteca para crear gráficos.
import matplotlib.pyplot as plt

# Importa NumPy para realizar operaciones con arreglos y datos.
import numpy as np

# np.unique() obtiene los valores únicos de la variable 'nivel_estudios'
# y cuenta cuántas veces aparece cada uno.
#
# - estudios: contiene las categorías únicas (por ejemplo: Primaria, Secundaria, Universidad).
# - frecuencia: contiene la cantidad de veces que aparece cada categoría.
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)

# Muestra en pantalla las categorías y sus respectivas frecuencias.
print(estudios, frecuencia)
```

```python
# DIAGRAMA DE BARRAS

# Etiqueta del eje Y (cantidad de observaciones).
plt.ylabel('Frecuencia absoluta')

# Etiqueta del eje X (categorías de la variable).
plt.xlabel('Nivel de estudios')

# Título del gráfico.
plt.title('Diagrama de barras')

# Crea el diagrama de barras.
#
# - estudios: categorías que aparecerán en el eje X.
# - frecuencia: altura de cada barra, correspondiente a la cantidad de datos.
plt.bar(estudios, frecuencia)
# Documentación:
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html

# Muestra el gráfico en pantalla.
plt.show()
```

```python
# DIAGRAMA DE SECTORES

# Define un conjunto de colores para cada sector del gráfico.
colores = {'red', 'blue', 'orange', 'green', 'cyan'}

# Crea un gráfico circular (pie chart).
#
# - frecuencia: determina el tamaño de cada sector.
# - labels=estudios: coloca el nombre de cada categoría.
# - colors=colores: asigna un color a cada sector.
# - autopct='%1.2f%%': muestra el porcentaje de cada categoría -> pct = percentaje
#   con dos decimales.
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')

# Título del gráfico.
plt.title('Diagrama de sectores')

# Muestra el gráfico en pantalla.
plt.show()
```

**Tarea: Gráficos en Pandas**

```python
#TABLA DE FRECUENCIA
#1ra Forma
import pandas as pd
'''
Crear un DataFrame a partir de un diccionario.
Cada clave del diccionario se convierte en el nombre de una columna,
mientras que los valores asociados serán los datos de dicha columna.
'''
dataframe = pd.DataFrame({
    "Nivel de estudios": estudios,
    "Frecuencia": frecuencia
})

# Mostrar el DataFrame en formato tabular.
display(dataframe)

# 2da forma
# Usando zip() combinamos dos listas o arreglos elemento por elemento.

# Posteriormente Pandas utiliza esas tuplas para crear las filas
# del DataFrame y el parámetro columns define el nombre de cada columna.
dataframe = pd.DataFrame(
    zip(estudios, frecuencia),
    columns=["Nivel de estudios", "Frecuencia"]
)

# Mostrar el DataFrame.
#display(dataframe)
```

```python
# DIAGRAMA DE BARRAS (Pandas)

# Crea un gráfico de barras utilizando los datos del DataFrame.
#
# x: columna que contiene las categorías.
# y: columna que contiene las frecuencias.
dataframe.plot(
    x="Nivel de estudios",
    y="Frecuencia",
    kind="bar",
    figsize=(8,5),
    legend=False
)

# Agrega un título al gráfico.
plt.title("Frecuencia por nivel de estudios")

# Etiqueta del eje X.
plt.xlabel("Nivel de estudios")

# Etiqueta del eje Y.
plt.ylabel("Frecuencia")

# Muestra una cuadrícula horizontal para facilitar la lectura.
plt.grid(axis="y")

# Muestra el gráfico.
plt.show()
```

```python
# DIAGRAMA DE PASTEL (Pandas)

# Crea un gráfico de pastel utilizando la columna de frecuencias.
#
# y: columna cuyos valores se representarán.
# labels: nombres de cada categoría.
dataframe.plot(
    y="Frecuencia",
    kind="pie",
    labels=dataframe["Nivel de estudios"],
    autopct="%1.1f%%",
    startangle=90,
    figsize=(7,7),
    legend=False
)

# Elimina la etiqueta del eje Y que Pandas agrega por defecto.
plt.ylabel("")

# Agrega un título.
plt.title("Distribución del nivel de estudios")

# Hace que el gráfico sea completamente circular.
plt.axis("equal")

# Muestra el gráfico.
plt.show()
```
